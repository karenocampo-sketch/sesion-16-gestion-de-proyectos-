### Cuello de botella y crecimiento profesional

### 1. Individual Contributor vs. Manager

Existen dos caminos principales para crecer en ingeniería:

* **Individual Contributor (IC):** crecer como especialista técnico, sin necesidad de convertirse en gerente.
* **Manager:** crecer liderando personas, equipos y procesos.

Ser un buen ingeniero no significa necesariamente convertirse en gerente.

### 2. La escalera corporativa

El crecimiento profesional **no depende únicamente de escribir buen código**.

Para subir de nivel también se tiene en cuenta:

* El impacto que generas en la organización.
* La capacidad de resolver problemas importantes.
* Cómo ayudas a otros ingenieros.
* Cómo mejoras procesos y sistemas.
* El impacto en costos e ingresos.
* La capacidad de justificar decisiones desde el punto de vista del negocio.

Por eso, al crecer, ya no solo importa **qué haces**, sino también **cómo lo haces y qué impacto genera**.

### 3. Nivel terminal

El **nivel terminal** es el nivel más alto al que una persona puede llegar dentro de una ruta determinada sin necesidad de pasar a otra.

Por ejemplo, llegar a un nivel como **L5** puede significar alcanzar un punto donde ya no basta con seguir haciendo más de lo mismo para ascender.

Para superar esa barrera se necesitan nuevas habilidades:

* Guiar a otros.
* Optimizar sistemas.
* Resolver problemas organizacionales.
* Entender las necesidades del negocio.
* Generar ahorro de costos o aumento de ingresos.
* Contribuir a la cultura de ingeniería.

### 4. El cambio de enfoque

A medida que aumenta la seniority, el trabajo deja de afectar solamente al proyecto propio y empieza a **impactar a toda la organización**.

Un senior debe pasar de:

**"Hago bien mi trabajo"**

a:

**"Hago que el sistema, el equipo y las personas funcionen mejor."**

### 5. Framework: Producir, Organizar y Publicar

**Producir:**
Crear soluciones, código, proyectos y resultados.

**Organizar:**
Coordinar personas, procesos, sistemas y recursos para que el trabajo funcione mejor.

**Publicar:**
Hacer visible el impacto del trabajo: documentar logros, compartir resultados y comunicar cómo el trabajo ayudó al equipo o al negocio.

No basta con hacer un buen trabajo: **el impacto debe ser visible y estar documentado.**

### 6. Liderazgo

Los líderes no solamente solucionan problemas. También son capaces de **identificar vacíos**:

* ¿Qué está faltando?
* ¿Qué proceso podría mejorar?
* ¿Dónde existe un problema que todavía nadie está atendiendo?
* ¿Qué necesita el equipo para avanzar?

### 7. Staff y niveles superiores

A partir de niveles como **Staff Engineer**, aumenta la importancia de la **persuasión interna**.

El ingeniero debe demostrar que su participación ayuda a que los sistemas sean:

* Más fáciles de mantener.
* Más organizados.
* Más escalables.
* Más eficientes.

Los proyectos también deben mostrar claramente el valor que generaron.

### 8. Promotion-Driven Development

El crecimiento no debería consistir únicamente en esperar una promoción.

La idea es empezar a desarrollar las habilidades y asumir responsabilidades del siguiente nivel **antes de llegar formalmente a él**.

Sin embargo, permanecer en el nivel terminal también puede ser una opción exitosa. No todos necesitan ascender continuamente.

### Idea principal

A medida que un ingeniero crece, su valor deja de estar solamente en **las líneas de código que produce** y pasa a estar también en su capacidad para **organizar personas, sistemas y necesidades del negocio**.

El objetivo es pasar de resolver problemas individuales a **hacer que toda la organización pueda resolver problemas de manera más efectiva**.
