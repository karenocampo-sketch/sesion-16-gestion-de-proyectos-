# Dueño de tu carrera

**Nadie se preocupa por tu carrera tanto como tú.**
Por eso, debes tomar el control de tu crecimiento profesional y no esperar a que otros decidan por ti.

### 1. Ser dueño de tu carrera

* Tener control sobre tu crecimiento profesional.
* Pensar como el **arquitecto de tu carrera**, no solo como alguien que programa o ejecuta tareas.
* Definir hacia dónde quieres crecer y qué habilidades necesitas.
* Muchos ingenieros se enfocan en el trabajo diario, pero no planean su carrera.

### 2. Retroalimentación

La retroalimentación debe basarse en **datos y hechos**, no en opiniones personales.

Se puede obtener información de:

* Revisiones de código.
* Documentos de diseño.
* Evaluaciones entre compañeros.
* Resultados y desempeño del trabajo.
* Feedback específico.

Al pedir feedback, es útil preguntar:

* **Ejemplos:** ¿Qué hice bien o qué debería mejorar?
* **Impacto:** ¿Cómo afectó mi trabajo al equipo o proyecto?
* **Sugerencias:** ¿Qué puedo hacer diferente?

Después, convertir el feedback en un **plan de acción concreto**.

### 3. Tener un gerente como aliado

El gerente puede ayudarte a conectar tu trabajo con los objetivos de la empresa.

Para construir una buena relación:

* Aprovechar las reuniones uno a uno, no solo para reportar problemas.
* Comunicar también los **logros y avances**.
* Entender los objetivos de tu gerente y de la organización.
* Generar confianza mutua.
* Resolver problemas antes de que se conviertan en emergencias.
* Documentar logros, decisiones y avances.
* Cumplir lo que prometemos.
* Si detectamos un cuello de botella, **levantar la mano y comunicarlo**.

### 4. Mantener un ritmo sostenible

No siempre trabajar al máximo significa ser más productivo.

* Los sistemas pueden fallar si funcionan constantemente al límite.
* El esfuerzo máximo consume mucha energía y puede llevar al agotamiento.
* Es mejor mantener un **ritmo constante y sostenible**.

### 5. Evitar el burnout

El objetivo es mantener un ritmo de trabajo que podamos sostener durante mucho tiempo, evitando llegar al agotamiento.

### 6. Evitar el "coasting"

**Coasting** significa trabajar por debajo de nuestra capacidad durante mucho tiempo.

* Puede hacer que produzcamos menos de lo esperado.
* Si se mantiene demasiado tiempo, podemos perder credibilidad profesional.
* La idea es encontrar un equilibrio entre **sobreexigirse y trabajar demasiado por debajo de nuestro potencial**.

### 7. Mantenimiento de la carrera

Así como un sistema necesita mantenimiento, nuestra carrera también.

Debemos revisar constantemente:

* Qué estamos aprendiendo.
* Qué habilidades necesitamos mejorar.
* En qué áreas queremos crecer.
* Qué nuevos retos podemos asumir.

La idea es mantener un equilibrio entre **crecimiento, aprendizaje y sostenibilidad**.
